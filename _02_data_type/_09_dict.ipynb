# dict
- 사전(dictionary) 자료형 
- 하나의 요소(item)는 key:value 쌍으로 존재

- 같은 dict 내에서 key는 중복될 수 없다. 
    - set은 순서 유지가 안된다. dict는 가능하다.(python 3.7+) 

- value는 중복 가능하다. 

- key는 hashable 이어야 한다. 
    - list, dict, set을 제외한 데이터 타입만 가능하다.
    - 보통 str을 가장 많이 사용한다. 

- dict는 lndex 대신 key를 이용해서 value를 조회한다. 



````python
# dict 생성
dct = {} # 빈 dict
print(type(dct))
````

````python
dct = {1}
print(type(dct))
````

````python
# dict 생성
dct = {} # 빈 dict
print(type(dct))

dct = {
    'a': 10,
    'b': 20,
    'c': 30
}

print(dct, type(dct))
````

````python
# tuple을 key로 지정
point_map = { (10, 20) : 'a 좌표' }
print(point_map)
````

## dict 조회

````python
dct = {
    'a': 10,
    'b': 20,
    'c': 30
}
````

````python
# [key] 를 이용한 조회
print("dct['a']:", dct['a'])
print("dct['b']:", dct['b'])
print("dct['c']:", dct['c'])
````

````python
# dict.get(key)를 이용한 조회
print("dct.get('a'):", dct.get('a'))
print("dct.get('b'):", dct.get('b'))
print("dct.get('c'):", dct.get('c'))
````

````python
# [], get() 차이 -> 없는 key를 이용한 조회

# [key]: key가 존재하지 않으면 KeyError 발생
# print("dct['x']:", dct['x']) 
````

````python
# get(): key가 존재하지 않으면 None 반환
print("dct.get('x'):", dct.get('x'))
````

## dict 요소(item == K:V) 추가

````python
students = {
    "s001" : "홍길동",
    "s002" : "신사임당",
    "s003" : "이순신"
}

# students에 item 추가
students['s004'] = '강감찬'

students.update(s005='일론머스크') # K=V  형식으로 작성

# update() 수행 시 dict를 이용해서 여러 item을 한 번에 추가 가능
students.update({'s006':'빌게이츠', 's007':'백동현'})

print(students)
````

## dict 요소(item == K:V) 제거

````python
# pop(key)을 이용한 item 제거
removed = students.pop('s001') # value만 반환된다
print(removed)
print(students) 
````

## dict 내장 함수

````python
# dict() 생성 함수
dct2 = dict(a=10, b=20)
print(dct2)

dct3 = dict([('name', '신사임당'), ('age', 33)])
print(dct3)
````

````python
# dict.keys() : dict key만 추출
keys = students.keys()
print(keys, type(keys))
````

````python
# dict.values() : dict에서 values만 추출
values = students.values()
print(values, type(values))
````

````python
# dict.items() : dict의 item(K:V)를 tuple로 묶어 반환
items = students.items()
print(items, type(items))
````

## dict unpacking

````python
# dict items unpacking
dct = dict(a=10, b=20, c=30)
print(dct)
print(dct.items())

for key, value in dct.items():
    print(key, value)
````

# dict 복사

### 값의 종류
- 원시 자료형(int, float, bool,,,)
    - 변수에 순수한 값 자체(리터럴)로 저장되는 자료형
- 참조 자료형(list, dict, set, str,,,)
    - 변수에 값이 저장된 위치(주소)를 참조(저장)하는 자료형
    - 따라서 참조 자료형을 대입하면 값 자체가 복사되는 것이 아니라 주소가 복사됨
    변수는 값 하나만 저장되기 때문에 리스트처럼 여러 값이 있는 건 넣을 수 없어 리스트 값이 있는 주소를 넣는다. 그 주소지에 있는 값들을 참조한다. 
    [1,2,3]에 대한 주소가 있고 이 주소를 참조한다.


### 대입
- A = B (A 변수에 B 값을 대입)
```

a = 10 # int(원시 자료형, 값 자체)
b = a # a에 저장되어있던 10이라는 값을 복사해서 b에 대입
# a = 10, b = 10

lst_a = [1,2,3] #list (참조 자료형, 값이 존재하는 곳의 주소)
lst_b = lst_a # list_a의 저장되어 있던 주소를 lst_b에 복사해서 대입

# lst_a와 lst_b가 같은 주소를 갖는다. == 같은 저장 공간을 공유한다. 

```

### 얕은 복사(shallow copy)
sample = { 'name': '기계식 키보드', 'price': 30\_000, 'origin': 'KR', }

# 얕은 복사: 주소만 복사하고 같은 저장 공간을 공유
other1 = sample


### 깊은 복사(deep copy): 복제의 개념
import copy
other2 = copy.deepcopy(sample)





````python
# 얕은 복사
sample = {
    'name': '기계식 키보드',
    'price': 30_000,
    'origin': 'KR',
}

other1 = sample # sample 변수에 저장된 주소를 other1에 복사'
# sample과 other1은 같은 곳을 바라보게 된다

print("sample:", sample)
print("other1:", other1)

# sample 값 변경
sample['price'] = 35_000

print("sample:", sample)
print("other1:", other1)
````

````python
# 깊은 복사
import copy

nested_sample = {
    'name': '노트북',
    'options': ['16GB RAM', '512GB SSD']
}

other2 = copy.deepcopy(nested_sample)

print("nested_sample:",  nested_sample)
print("other2:",  other2)

# nested_sample 변경
nested_sample['name'] = '데스크탑'

print("nested_sample:",  nested_sample)
print("other2:",  other2)
````

````python
# 깊은 복사는 원본이나 복사본을 변경해도 서로 영향을 주지 않음
# 얕은 복사는 원본이 따로 있어서 원본 변경 시 복사본이 영향을 받음
````
